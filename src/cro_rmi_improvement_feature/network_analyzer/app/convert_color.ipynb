```python
def hex_to_rgb(hex_colors):
    """
    Converts a list of hexadecimal color codes to a list of RGB tuples.

    Args:
        hex_colors: A list of strings, where each string is a hex color code
                    (e.g., '#RRGGBB').

    Returns:
        A list of RGB tuples, where each tuple contains three integers
        (red, green, blue).
    """
    rgb_colors = []
    for hex_color in hex_colors:
        # Remove the '#' from the string
        hex_color = hex_color.lstrip("#")
        # Convert hex parts to integers
        r = int(hex_color[0:2], 16)
        g = int(hex_color[2:4], 16)
        b = int(hex_color[4:6], 16)
        rgb_colors.append((r, g, b))
    return rgb_colors


hex_list = [
    "#e6194B",
    "#3cb44b",
    "#ffe119",
    "#4363d8",
    "#f58231",
    "#911eb4",
    "#42d4f4",
    "#f032e6",
    "#bfef45",
    "#fabed4",
    "#469990",
    "#dcbeff",
    "#9A6324",
    "#fffac8",
    "#800000",
    "#aaffc3",
    "#808000",
    "#ffd8b1",
    "#000075",
    "#a9a9a9",
]
rgb_list = hex_to_rgb(hex_list)
rgb_str_list = [f"rgb({r}, {g}, {b})" for r, g, b in rgb_list]
print(rgb_str_list)
```
