```python
direction_set = []

for company_id in data["company_graph_datas"].keys():
    if "oneway_run" in company_id:
        company_data = data["company_graph_datas"][company_id]
        company_data_edges = company_data["edges"]
        for edge in company_data_edges:
            direction = edge["direction"]
            if direction not in direction_set:
                direction_set.append(direction)
direction_set
```
