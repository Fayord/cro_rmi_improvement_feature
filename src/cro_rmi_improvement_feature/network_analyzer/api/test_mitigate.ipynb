```python
import json

data_path = "example_request.json"
with open(data_path, "r") as f:
    data = json.load(f)

data.keys()
data["data_level"] = "company"
```

```python
import requests

response = requests.post(url, json=data)
total_time = response.elapsed.total_seconds()
response.status_code
print(f"Total time: {total_time} seconds")
print(f"Status code: {response.status_code}")
```

```python
result = response.json()
print(result.keys())
recommendations = result["recommendations_with_tags"]
print(len(recommendations))
for recommendation in recommendations:
    print(recommendation.keys())
    print(recommendation["risk_data"])
    print("source_risk" if recommendation["is_source_risk"] else "")
    print("central_risk" if recommendation["is_central_risk"] else "")
    print("high_risk" if recommendation["is_high_risk"] else "")
    print("shared_root_cause" if recommendation["is_shared_root_cause"] else "")
    print("news_trended" if recommendation["is_news_trended"] else "")
    print("emerging_risk" if recommendation["is_emerging_risk"] else "")
    print()
```
