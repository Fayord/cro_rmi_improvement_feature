```python
import requests

# url = "http://0.0.0.0:7800/rmi_graph_based_recommendation/recommend_risk_to_assess"
url = (
    "http://172.16.100.50:7900/rmi_graph_based_recommendation/recommend_risk_to_assess"
)
data = {
    "id": "PCG|embedding_summary_user_data|oneway_run",
    # "id": "test",
    "data_level": "user",
    "year_quarter": "2025-Q1",
    "existing_risks": [
        {
            "risk_id": "risk_001",
            "user_id": "user_001",
            "company_id": "company_001",
            "country": "country_001",
            "risk_name": "Supply shortage",
            "risk_description": "Risk of disruption in supply chain operations",
            "processes": [
                {
                    "id": "1",
                    "name": "Procurement",
                    "description": "Vendor selection and management process",
                }
            ],
            "root_causes": [
                {
                    "id": "1",
                    "name": "Single source dependency",
                    "description": "Over-reliance on single supplier",
                }
            ],
            "risk_category": "operational",
            "score": {
                "impact": 4,
                "likelihood": 3,
                "score": 12,
                "risk_level": 3,
            },
        }
    ],
}

response = requests.post(url, json=data)
response.status_code
```

```python
result = response.json()
recommendations = result["recommendations"]
print(len(recommendations))
for recommendation in recommendations:
    print(recommendation.keys())
    print(recommendation["risk"])
    print()
```

```python
import sys
import os

dir_path = os.getcwd()
sys.path.append(f"{dir_path}/../")
sys.path.append(f"{dir_path}/../../find_similar_risk")

from embedding_providers import (  # type: ignore
    OpenAIEmbeddingProvider,
)
from typing import List


# in here use use_cache = False to see the time
def create_embedding(text: str, use_cache: bool = False) -> List[float]:
    provider = OpenAIEmbeddingProvider(model_name="text-embedding-3-large")

    def get_openai_large_embedding(text: str) -> List[float]:
        """Get embedding from OpenAI large model."""
        if not text or text.strip() == "":
            return None
        try:
            embedding = provider.get_embedding(text, use_cache=use_cache)
            return embedding.tolist()
        except Exception as e:
            print(f"Error getting embedding for text: {e}")
            return None

    return get_openai_large_embedding(text)


example_text = "Risk: Supply Chain Disruption | Description: Risk of disruption in supply chain operations | Processes: Procurement: Vendor selection and management process | Root Causes: Single source dependency: Over-reliance on single supplier"
```
