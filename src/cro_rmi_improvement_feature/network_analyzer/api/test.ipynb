```python
import requests

url = "http://0.0.0.0:7900/rmi_graph_based_recommendation/recommend_risk_to_assess"
data = {
    "company_id": "PCG|embedding_summary_user_data",
    "existing_risks": [
        {
            "risk_id": "risk_001",
            "risk_name": "Supply Chain Disruption",
            "risk_description": "Risk of disruption in supply chain operations",
            "processes": [
                {
                    "id": 1,
                    "name": "Procurement",
                    "description": "Vendor selection and management process",
                }
            ],
            "root_causes": [
                {
                    "id": 1,
                    "name": "Single source dependency",
                    "description": "Over-reliance on single supplier",
                }
            ],
            "risk_category": "operational",
            "score": {
                "impact": 4,
                "likelihood": 3,
                "score": 12,
                "risk_level": 3,
            },
        }
    ],
}

response = requests.post(url, json=data)
response.status_code
```

```python
result = response.json()
recommendations = result["recommendations"]
print(len(recommendations))
for recommendation in recommendations:
    print(recommendation["risk_name"])
    print(recommendation.keys())
    recommend_risk_list = recommendation["recommend_risk_list"]
    print(len(recommend_risk_list))
    for risk in recommend_risk_list:
        print(risk["risk"])
```
