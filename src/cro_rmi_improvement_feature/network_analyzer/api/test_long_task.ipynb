```python
import requests
import time
import json

# The base URL of your FastAPI application
base_url = "http://127.0.0.1:8000"

# --- Test Case 1: Start a long-running task with the correct data format ---
print("Step 1: Starting the processing task...")
# The payload needs to be a dictionary inside the 'data' key.
correct_data_to_process = {"data": {"payload": "some_heavy_payload"}}
response = requests.post(f"{base_url}/api/process", json=correct_data_to_process)
task_info = response.json()
print(f"Response from /api/process: {task_info}")
task_id = task_info.get("task_id")

if task_id:
    print(f"Task ID received: {task_id}")
    print("-" * 30)

    # --- Test Case 2: Check status immediately (should be "in_progress") ---
    print("Step 2: Checking status immediately...")
    status_response_1 = requests.get(f"{base_url}/api/results/{task_id}")
    status_info_1 = status_response_1.json()
    print(f"Status check response (initial): {status_info_1}")
    print("-" * 30)

    # --- Test Case 3: Wait for a few seconds and check again ---
    print("Step 3: Waiting for 12 seconds for the task to complete...")
    for i in range(12):
        time.sleep(1)
        print(f"Waiting for {i+1} seconds...")

        status_response_2 = requests.get(f"{base_url}/api/results/{task_id}")
        status_info_2 = status_response_2.json()
        print(f"Status check response (after wait): {status_info_2}")
        # status has to be in_progress, success, or fail
        if status_info_2.get("status") == "success":
            print("\n✅ Test passed: The task completed successfully!")
            print(f"Final data: {status_info_2.get('data')}")
            break
        elif status_info_2.get("status") == "fail":
            print("\n❌ Test failed: The task did not complete successfully.")
            print(f"Error: {status_info_2.get('error')}")
            break
        elif status_info_2.get("status") == "in_progress":
            print(f"Status: {status_info_2.get('status')}")
            print(f"Message: {status_info_2.get('message')}")
        else:
            print("got other status", status_info_2.get("status"))
            break

# The rest of the tests for the non-existent task ID remain correct.
```
